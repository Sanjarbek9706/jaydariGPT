```python
print('Hello world!')
# Bu darsda so'zlarni token deb olib, sodda misol tuzamiz.
# Haqiqiy AI tokenizerida 1 ta so'z doim 1 ta token bo'lmaydi.
```

```python
# text o'zgaruvchisida ishlov beriladigan matnni saqlaymiz.
# Bu matnda bo'shliq bo'yicha ajratilganda 17 ta so'z bor.
text = 'Hello world this is a simple example to show how tokenization works in NLP and a sentence'
# Saqlangan matnni ekranga chiqaramiz.
print(text)
```

```python
# split() matnni bo'shliqlar bo'yicha ajratib, ro'yxat (list) qaytaradi.
# Bu sodda so'zlarga ajratish misoli; tiktoken boshqa qoidalardan foydalanadi.
tokens = text.split()
# Ajratilgan 17 ta elementni ko'ramiz; 'a' ikki marta uchraydi.
print(tokens)
```

```python
# Har bir takrorlanmaydigan so'zga raqam berish uchun bo'sh lug'at yaratamiz.
vocab = {}
# Birinchi so'zning ID raqami 0 dan boshlanadi.
current_id = 0

# Ro'yxatdagi har bir so'zni ketma-ket tekshiramiz.
for token in tokens:
    # So'z lug'atda yo'q bo'lsagina unga yangi ID beramiz.
    if token not in vocab:
        vocab[token] = current_id
        # Keyingi yangi so'z uchun ID ni bittaga oshiramiz.
        current_id += 1

# 17 ta so'zdan 16 tasi noyob: 'a' takrorlangani uchun ID si o'zgarmaydi.
print(vocab)
```

```python
# Har bir so'z o'rniga lug'atdagi ID sini olib, yangi ro'yxat tuzamiz.
# Bu qisqa yozuv list comprehension deyiladi.
encoded = [vocab[token] for token in tokens]
# Masalan, 'Hello' -> 0, 'world' -> 1, 'a' -> 4.
# 'a' ikki joyda kelgani sababli 4 raqami ham ikki joyda keladi.
print(encoded)
```

```python
# Lug'at yo'nalishini teskariga o'giramiz: so'z -> ID o'rniga ID -> so'z.
# items() lug'atdagi (so'z, ID) juftliklarini beradi.
reverse_vocab = {id: token for token, id in vocab.items()}
# Har bir ID dan so'zni topamiz; join() ularni bitta bo'shliq bilan birlashtiradi.
decoded = " ".join([reverse_vocab[id] for id in encoded])
# Ushbu misolda dastlabki matn tiklanadi.
# split()/join() asl matndagi ortiqcha bo'shliqlarni aynan saqlamaydi.
print(decoded)
```

```python
# Embedding vektori: tokenni ifodalovchi sonlar ro'yxati.
# Quyidagi sonlar faqat tushuntirish uchun yozilgan misollar:
# 0 => [0.41, 0.22, -0.77, 0.56, ...]
# 1 => [0.34, 0.72, 0.77, 0.82, ...]
# Token ID — bitta raqam; embedding esa ko'p sonli vektor.
# Vektorlar o'xshashlik bo'yicha qidirishda vektor bazasida saqlanishi mumkin.
```

```python
# PyTorch: tensorlar va neyron tarmoq qatlamlari bilan ishlash kutubxonasi.
import torch
import torch.nn as nn

# Har bir token 8 ta sondan iborat vektor bilan ifodalanadi.
embedding_dim = 8
# Lug'atdagi takrorlanmaydigan so'zlar soni: bu misolda 16 ta.
vocab_size = len(vocab)

# Har bir token ID uchun vektor saqlaydigan o'rgatiladigan jadval yaratamiz.
# Dastlab vektorlar tasodifiy: ular hali so'z ma'nosini o'rganmagan.
embedding_layer = nn.Embedding(vocab_size, embedding_dim)

# ID lar ro'yxatini PyTorch tensoriga aylantiramiz.
# Tashqi [] bitta matndan iborat guruh (batch) hosil qiladi: [1, 17].
input_ids = torch.tensor([encoded])

# Har bir ID uchun jadvaldagi 8 o'lchamli vektorni olamiz.
# Takrorlangan ID, masalan 'a', aynan bir xil vektorni oladi.
embeddings = embedding_layer(input_ids)

# Barcha vektorlarni ekranga chiqaramiz.
print(embeddings)
# [1, 17, 8]: 1 ta matn, 17 ta token, har token uchun 8 ta son.
print(embeddings.shape)
```

```python
# | Bosqich | Esda qoladigan ma’nosi |
# |---|---|
# | **Text** | Siz yozgan gap |
# | **Tokenization** | Gapni bo‘laklash |
# | **Token ID** | Har bo‘lakning lug‘atdagi raqami |
# | **Embedding vectors** | Har raqamni sonlar vektori bilan ifodalash |
# | **Birinchi Transformer block** | Bog‘lanishlarni hisoblab, vektorlarni yangilash |
# | **Keyingi Transformer block** | Oldingi natijani olib, yana qayta ishlash |
# | **Chiqish bosqichi** | Keyingi tokenni tanlash va matnni davom ettirish |
```

```python
# Hugging Face tokenizerlari va modellaridan foydalanish kutubxonasini o'rnatadi.
# ! belgisi Colab'da terminal buyrug'ini bajaradi; pip paket o'rnatadi.
!pip install transformers
```

```python
# Modelga mos tokenizerni avtomatik tanlaydigan klassni import qilamiz.
from transformers import AutoTokenizer

# Hugging Face'dagi model manzili: tashkilot nomi / model nomi.
model_id = "Qwen/Qwen3-1.7B"
# Tayyor tokenizer lug'ati va ajratish qoidalarini yuklaymiz.
# Bu qator modelning javob yaratuvchi og'irliklarini yuklamaydi.
tokenizer = AutoTokenizer.from_pretrained(model_id)
```

```python
# Tokenizerning asosiy lug'atidagi tokenlar sonini ko'rsatadi.
# Qo'shimcha tokenlar bo'lsa, len(tokenizer) bundan farq qilishi mumkin.
print("Vocab size:", tokenizer.vocab_size)
# Maxsus token nomlari va belgilarini ko'rsatadi.
# eos_token — tugash; pad_token — matn uzunligini tenglashtirish belgisi.
print('Special tokens:', tokenizer.special_tokens_map)
```

```python
# Tokenlarga ajratish uchun namuna matnni saqlaymiz.
text = "Hello how are you doing?"
# Asl matnni chiqaramiz; keyin uni tokenlar va ID lar bilan solishtiramiz.
print('text:', text)
```

```python
# Matnni tanlangan tokenizer qoidalariga ko'ra bo'laklarga ajratamiz.
# Bu usul sonli ID emas, tokenlarning yozuvini qaytaradi.
tokens = tokenizer.tokenize(text)
# Ushbu Qwen tokenizerida Ġ belgisi token oldidagi bo'shliqni bildiradi.
# Masalan, Ġhow asl matndagi ' how' bo'lagini ifodalaydi.
print('tokens:', tokens)
```

```python
# Oldingi katakdagi har bir tokenni lug'atdagi ID raqamiga aylantiramiz.
# convert_tokens_to_ids() matnni qayta bo'laklamaydi: tayyor tokenlarni qabul qiladi.
ids = tokenizer.convert_tokens_to_ids(tokens)
# ID — tokenning lug'atdagi raqami; uning ma'nosi yoki muhimlik balli emas.
print('ids:', ids)
```

```python
# Tokenizerni chaqirib, matnni modelga berishga tayyor ko'rinishga keltiramiz.
encoded = tokenizer(text)
# input_ids — token ID lari.
# attention_mask — haqiqiy tokenlar uchun 1, padding joylari uchun 0.
# Bu yerda padding yo'q, shu sabab maskadagi qiymatlar 1 bo'ladi.
print('encoded:', encoded)
```

```python
# encoded natijasidan faqat input_ids maydonini tanlab chiqaramiz.
# ['input_ids'] lug'atdan shu nomdagi qiymatni olishni bildiradi.
print('input ids:', encoded['input_ids'])
```

```python
# Attention maskni tushunish uchun shartli misol; ID lar haqiqiy hisob emas.
'Hi' # => [1, 345] — shartli uzunlik: 2 ta token.
'How are you?' # => [1, 45, 567, 234, 23453] — shartli uzunlik: 5 ta token.
# Qisqa ketma-ketlik 5 taga tenglashtirilsa, unga 3 ta padding qo'shiladi.
# Shunda uning attention_mask qiymati: [1, 1, 0, 0, 0].
# 1 — haqiqiy tokenni hisobga ol; 0 — padding joyini e'tiborsiz qoldir.
# Padding ID raqami esa tanlangan tokenizerga bog'liq.
```

```python
# Token ID larini yana o'qiladigan matnga aylantiramiz: bu decoding deyiladi.
decoded_text = tokenizer.decode(encoded['input_ids'])
# Qayta tiklangan matnni chiqarib, boshlang'ich matn bilan solishtiramiz.
print('decoded text:', decoded_text)
# Agar maxsus tokenlarni yashirmoqchi bo'lsak, decode() ga
# skip_special_tokens=True parametrini qo'shish mumkin.
```

```python
# Uchta matnni birgalikda tokenlarga aylantiramiz.
texts = [
    "Hello, how are you?",
    "Today we are learning about tokenizers.",
    "TinyLlama is a small but powerful language model."
]

# padding=True: qisqa matnlarni maxsus token bilan teng uzunlikka keltiradi.
# return_tensors="pt": natijani PyTorch tensorlari ko'rinishida qaytaradi.
batch = tokenizer(texts, padding=True, return_tensors="pt")
# input_ids — token raqamlari; attention_mask — haqiqiy token: 1, padding: 0.
print(batch.keys())

# shape: [matnlar soni, paddingdan keyingi tokenlar soni].
print("Input IDs shape:", batch["input_ids"].shape)
print("Attention mask shape:", batch["attention_mask"].shape)
```

```python
# Bitta uzun so'z ham bir nechta tokenga bo'linishi mumkin.
word = "headerandshoulder"
# Asl so'zni chiqaramiz.
print("word:", word)
# So'zning tokenlarga qanday ajratilganini ko'ramiz.
print("Tokens:", tokenizer.tokenize(word))
# encode() matnni to'g'ridan-to'g'ri token ID lariga aylantiradi.
print("IDs:", tokenizer.encode(word))
```
